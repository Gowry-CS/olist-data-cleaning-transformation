Since products dataframe has 2 more product category name than product_cat dataframe, we will do an outer join. 
Following that, we can look into the additional categories that have no translation. 

```python
# Outer join both dataframes
outer_merged_prod = pd.merge(products, product_cat, on='product_category_name', how='outer')

print(outer_merged_prod)
```

```python
# number of null values in outer_merged
print(outer_merged_prod.isna().sum())
```

```python
def normalize_name(text):
    if pd.isna(text):  # keep NaN safe
        return text
    
    # lowercase + strip spaces
    text = text.lower().strip()
    
    # collapse multiple spaces into one
    text = re.sub(r'\s+', ' ', text)
    
    # remove special characters like '-' or '/'
    text = text = re.sub(r'[^a-zA-Z0-9]', ' ', text)
    
    return text
```

```python
outer_merged_prod['product_category_name'] = outer_merged_prod['product_category_name'].apply(normalize_name)
print(outer_merged_prod)
```

```python
# Get rows that have no product translation to english

no_translation = outer_merged_prod['product_category_name_english'].isna()
bad_rows = outer_merged_prod[no_translation]
result = bad_rows[['product_category_name','product_category_name_english']].drop_duplicates() 
print(result)
```

```python
# Initialize the Translator
translator = Translator()

# Define function to translate text
def translate_to_english(text):
    """Translate a Portuguese string to English using Google Translate."""
    if pd.isna(text):   # Handle NaN safely
        return None
    try:
        translation = translator.translate(str(text), src='pt', dest='en')
        return translation.text
    except Exception as e:
        print(f"Error translating '{text}': {e}")
        return None
```

```python
result['google_trans_english'] = result['product_category_name'].apply(translate_to_english)
print(result)
```

```python
# Drop product_category_name_english from result
result.drop('product_category_name_english', axis=1, inplace=True)
print(result)
```

```python
outer_merged_prod = outer_merged_prod.merge(result, on='product_category_name', how='left')
print (outer_merged_prod)
```

```python
# Update the missing values to product_category_name_english
outer_merged_prod['product_category_name_english'] = (outer_merged_prod['product_category_name_english']).fillna(outer_merged_prod['google_trans_english'])
print(outer_merged_prod.isna().sum())
```

```python
# Drop google_trans_english column
outer_merged_prod = outer_merged_prod.drop(columns=['google_trans_english'])
```
