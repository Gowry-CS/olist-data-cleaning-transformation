##### Create Fact & Dimension Tables for Gold Lakehouse

```python
# Customers Dimension Table
spark.sql("DROP TABLE IF EXISTS GoldLakehouse.dim_customers")

spark.sql("""
CREATE TABLE GoldLakehouse.dim_customers (
    customer_id STRING,
    customer_unique_id STRING,
    customer_zip_code_prefix STRING,
    customer_city STRING,
    customer_state STRING
)
USING DELTA
""")
```

```python
# Sellers Dimension Table
spark.sql("DROP TABLE IF EXISTS GoldLakehouse.dim_sellers")

spark.sql("""
CREATE TABLE GoldLakehouse.dim_sellers (
    seller_id STRING,
    seller_zip_code_prefix STRING,
    seller_city STRING,
    seller_state STRING
)
USING DELTA
""")
```

```python
# Product Dimension Table
spark.sql("DROP TABLE IF EXISTS GoldLakehouse.dim_products")

spark.sql("""
CREATE TABLE GoldLakehouse.dim_products (
    product_id STRING,
    product_category_name STRING,
    product_name_length INT,
    product_description_length INT,
    product_photos_qty INT,
    product_weight_g FLOAT,
    product_length_cm FLOAT,
    product_height_cm FLOAT,
    product_width_cm FLOAT
)
USING DELTA
""")
```

```python
# Date Dimension Table
spark.sql("DROP TABLE IF EXISTS GoldLakehouse.dim_dates")

spark.sql("""
CREATE TABLE GoldLakehouse.dim_dates (
    date_id DATE,
    day INT,
    month INT,
    year INT,
    quarter INT,
    day_of_week STRING
)
USING DELTA
""")
```

```python
# Orders Fact Table
spark.sql("DROP TABLE IF EXISTS GoldLakehouse.fact_orders")

spark.sql("""
CREATE TABLE GoldLakehouse.fact_orders (
    order_id STRING,
    customer_id STRING,
    order_status STRING,
    order_purchase_timestamp TIMESTAMP,
    order_approved_at TIMESTAMP,
    order_delivered_carrier_date TIMESTAMP,
    order_delivered_customer_date TIMESTAMP,
    order_estimated_delivery_date TIMESTAMP,
    total_order_value FLOAT,
    freight_value FLOAT,
    review_score INT,
    date_id DATE
)
USING DELTA
""")
```

```python
# Order Items Table
spark.sql("DROP TABLE IF EXISTS GoldLakehouse.fact_order_items")

spark.sql("""
CREATE TABLE GoldLakehouse.fact_order_items (
    order_item_id INT,     
    order_id STRING,
    product_id STRING,
    seller_id STRING,
    shipping_limit_date TIMESTAMP,
    price FLOAT,
    freight_value FLOAT,
    date_id DATE
)
USING DELTA
""")
```

```python
# Payments Fact Table
spark.sql("DROP TABLE IF EXISTS GoldLakehouse.fact_payments")

spark.sql("""
CREATE TABLE GoldLakehouse.fact_payments (
    order_id STRING,
    customer_id STRING,
    payment_sequential INT,
    payment_type STRING,
    payment_installments INT,
    payment_value FLOAT,
    date_id DATE
)
USING DELTA
""")
```

```python
# Reviews Fact Table
spark.sql("DROP TABLE IF EXISTS GoldLakehouse.fact_reviews")

spark.sql("""
CREATE TABLE GoldLakehouse.fact_reviews (
    review_id STRING,
    order_id STRING,
    customer_id STRING,
    review_score INT,
    order_status STRING,
    review_comment_title STRING,
    review_comment_message STRING,
    review_creation_date DATE,
    review_answer_timestamp TIMESTAMP
)
USING DELTA
""")
```

```python
# Create Geolocation Dimension
spark.sql("DROP TABLE IF EXISTS GoldLakehouse.dim_geolocation")

spark.sql("""
CREATE TABLE GoldLakehouse.dim_geolocation (
    geolocation_zip_code_prefix STRING,
    geolocation_city STRING,
    geolocation_state STRING,
    geolocation_lat DOUBLE,
    geolocation_lng DOUBLE
)
USING DELTA
""")
```
